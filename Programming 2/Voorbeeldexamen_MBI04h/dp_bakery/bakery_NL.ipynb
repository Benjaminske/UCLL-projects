# Data processing: Bakkerij verkoopsdata


## Over de dataset

Deze dataset bevat de dagelijkse transactiegegevens van klanten van een Franse bakkerij in de periode van 2021-01-01 tot 2022-09-30.


**Inhoud**
De dataset bevat 234005 rijen, verspreid over 136000 transacties en 6 kolommen.

**Variabelen**

* `date`: datum van de bestelling
* `time`: tijdstip van de bestelling
* `ticket_number`: identificatie voor elke afzonderlijke transactie; 1 ticket_number = één bestelling
* `article`: naam van het verkochte product (in het Frans)
* `quantity`: hoeveel producten er zijn verkocht
* `unit_price`: prijs per product
* `retour`: of een gegeven rij deel uitmaakt van een retour of iet
  * *False*: de producten zijn verkocht, niet geretourneerd
  * *True*: dit is een retour
  * *product_number*: deze producten zijn geretourneerd, het product_number van de retour wordt vermeld
  * *partial*: deze producten zijn gedeeltelijk geretourneerd

**Opmerking**

* Het product COUPE wordt gebruikt wanneer klanten vragen om hun brood te laten snijden (met de machine).
